```python
import numpy as np

np.random.seed(42)

p = 10
theta0 = np.ones(p) * 6.0

def L(theta):
    return (np.dot(theta, theta) / 5 - 25) ** 2

def y_a(theta):
    return L(theta) + np.random.normal(0, 2)

def y_b(theta):
    V = np.random.normal(0, 0.1, size=11)
    return L(theta) + np.dot(np.append(theta, 1), V)

def run_spsa(y_func, a, A, c, n_meas=1000):
    theta = theta0.copy()
    n_iter = n_meas // 2

    for k in range(n_iter):
        ak = a / (k + 1 + A) ** 0.602
        ck = c / (k + 1) ** 0.101

        delta = 2 * np.random.binomial(1, 0.5, size=p) - 1

        y_plus = y_func(theta + ck * delta)
        y_minus = y_func(theta - ck * delta)

        ghat = (y_plus - y_minus) / (2 * ck * delta)
        theta = theta - ak * ghat

    return theta

# Part (a)
np.random.seed(42)
th_a = run_spsa(y_a, a=0.00164, A=50, c=2.0)

# Part (b)
np.random.seed(43)
th_b = run_spsa(y_b, a=0.00166, A=50, c=1.9)

print('--- Part (a): eps ~ N(0, 4) ---')
print(f'L(terminal)  = {L(th_a):.4f}')
print(f'theta^T theta = {np.dot(th_a, th_a):.4f}')
print()

print('--- Part (b): eps = [theta^T, 1]V ---')
print(f'L(terminal)  = {L(th_b):.4f}')
print(f'theta^T theta = {np.dot(th_b, th_b):.4f}')
print()

print(f'L(theta_0)   = {L(theta0)}')
print('L(theta*)    = 0  (for any theta* with theta^T theta = 125)')
```
