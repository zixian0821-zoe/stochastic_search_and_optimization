```python
import numpy as np

P_c = 0.70
P_m = 0.005
B   = 50

lower_bound = (1 - P_c) * (1 - P_m) ** B
print("lower bound =", round(lower_bound, 4))

# check with simulation
rng = np.random.default_rng(0)
n = 1_000_000
parent = np.zeros(B, dtype=int)
count = 0
for _ in range(n):
    child = parent.copy()
    if rng.random() < P_c:
        mate = rng.integers(0, 2, B)
        k = rng.integers(1, B)
        child[k:] = mate[k:]
    flips = rng.random(B) < P_m
    child = np.where(flips, 1 - child, child)
    if child.sum() == 0:
        count += 1
print("simulation =", round(count / n, 4))
```

```python
import itertools
from math import comb

N = 2
B = 2

chroms = list(itertools.product([0, 1], repeat=B))

# (a)
ordered = list(itertools.product(chroms, repeat=N))
print("(a) ordered populations =", len(ordered), " (formula 2^NB =", 2**(N*B), ")")

# (b)
unique = sorted({tuple(sorted(p)) for p in ordered})
print("(b) unique populations =", len(unique), " (formula C(N+2^B-1, N) =", comb(N + 2**B - 1, N), ")")
for p in unique:
    print("   ", p)

# (c)
absorbing = [p for p in unique if p[0] == p[1]]
print("(c) absorbing states =", len(absorbing))
for p in absorbing:
    print("   ", p)
```
